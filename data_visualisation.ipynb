```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
pd.set_option('display.max_columns', 100)
```

```python
mc_20_25 = pd.read_pickle('Results/mc_RMT_7_noise_25.pickle')
summary_7_25 = pd.read_pickle('Results/summary_RMT_7_noise_25.pickle')

mc_7_75 = pd.read_pickle('Results/mc_RMT_7_noise_75.pickle')
summary_7_75 = pd.read_pickle('Results/summary_RMT_7_noise_75.pickle')

mc_20_25 = pd.read_pickle('Results/mc_RMT_20_noise_25.pickle')
summary_20_25 = pd.read_pickle('Results/summary_RMT_20_noise_25.pickle')

mc_20_75 = pd.read_pickle('Results/mc_RMT_20_noise_75.pickle')
summary_20_75 = pd.read_pickle('Results/summary_RMT_20_noise_75.pickle')

def plot_box_plot(mc,title):
    plot_metrics = ['Ann. vol (%)','CAGR (%)','Sharpe','Sortino','Max DD (%)']          # pick the perf_stats keys you care about
    fig, axes = plt.subplots(1, len(plot_metrics), figsize=(4.5 * len(plot_metrics), 4))
    for ax, m in zip(axes, plot_metrics):
        mc.pivot(index='iter', columns='method', values=m)[order].boxplot(ax=ax, rot=45,medianprops=dict(color='black', linewidth=2), showfliers=False)
        ax.set_title(m)
    plt.suptitle(title)
    plt.tight_layout()
    plt.show()

def summary_stats(mc):
    metrics = [c for c in mc.columns if c not in ('method', 'iter')]
    def q(p): return (f'p{int(p*100):02d}', lambda x: x.quantile(p))
    iqr = ('IQR', lambda x: x.quantile(.75) - x.quantile(.25))
    summary = mc.groupby('method')[metrics].agg([q(.05), q(.5), q(.95), iqr]).loc[order]
    display(summary.round(3))

plot_box_plot(mc_7_25,'Boxplot for 7 strategies with 25 noise')
summary_stats(mc_7_25)
plot_box_plot(mc_7_75,'Boxplot for 7 strategies with 75 noise')
summary_stats(mc_7_75)
plot_box_plot(mc_20_25,'Boxplot for 20 strategies with 25 noise')
summary_stats(mc_20_25)
plot_box_plot(mc_20_75,'Boxplot for 20 strategies with 75 noise')
summary_stats(mc_20_75)
```

```python
from scipy import stats

# True = higher is better. Check Max DD's sign convention in your perf_stats:
# if it's stored as negative (e.g. -25), higher is better; if it's a positive magnitude, set False.
higher_is_better = {'Ann. vol (%)': False, 'CAGR (%)': True, 'Sharpe': True,
                    'Max DD (%)': True, 'Calmar': True}

def test_vs_ew(mc, metrics=higher_is_better, base='EW'):
    rows = []
    for m, hib in metrics.items():
        P = mc.pivot(index='iter', columns='method', values=m)
        for k in P.columns.drop(base):
            d = (P[k] - P[base]).replace([np.inf, -np.inf], np.nan).dropna()
            if not hib:
                d = -d                                       # so positive always = better than EW
            wins, n = int((d > 0).sum()), int((d != 0).sum())
            rows.append(dict(
                metric=m, method=k,
                mean_better=d.mean(), median_better=d.median(),
                win_rate=wins / len(d),
                effect_size=d.mean() / d.std(),              # Cohen's d for paired data
                p_t=stats.ttest_1samp(d, 0).pvalue,
                p_wilcoxon=stats.wilcoxon(d).pvalue,
                p_sign=stats.binomtest(wins, n, 0.5).pvalue,
            ))
    out = pd.DataFrame(rows)
    out['p_wilcoxon_adj'] = stats.false_discovery_control(out['p_wilcoxon'])   # multiple-testing correction
    return out

print('20 Strategies with 25 Noise')
res_20_25 = test_vs_ew(mc_20_25)
display(res_20_25.pivot(index='method', columns='metric', values='p_wilcoxon_adj').round(4))
display(res_20_25.pivot(index='method', columns='metric', values='effect_size').round(2))

print('7 Strategies with 25 Noise')
res_7_25 = test_vs_ew(mc_7_25)
display(res_7_25.pivot(index='method', columns='metric', values='p_wilcoxon_adj').round(4))
display(res_7_25.pivot(index='method', columns='metric', values='effect_size').round(2))
```
